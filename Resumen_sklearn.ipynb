```python
# @title
# ── Librerías generales ───────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import warnings
warnings.filterwarnings("ignore")

# ── Scikit-learn: lo que vamos a usar ─────────────────────────────────
from sklearn import datasets

# Preprocesamiento
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.impute import SimpleImputer

# División de datos
from sklearn.model_selection import (
    train_test_split, cross_val_score,
    GridSearchCV, learning_curve, StratifiedKFold
)

# Clasificación
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier, GradientBoostingClassifier,
    AdaBoostClassifier, VotingClassifier
)
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# Regresión
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# Clustering
from sklearn.cluster import KMeans, DBSCAN
from sklearn.mixture import GaussianMixture

# Reducción de dimensionalidad
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Métricas
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    roc_auc_score, roc_curve, precision_recall_curve,
    mean_squared_error, r2_score, silhouette_score
)

# Features y pipeline
from sklearn.feature_selection import SelectKBest, f_classif, RFE
from sklearn.pipeline import Pipeline

# ── Estilo visual de las gráficas ─────────────────────────────────────
# Paleta inspirada en Prometheus: azules profundos, cian eléctrico, naranja
CYAN   = "#00d4ff"
BLUE   = "#1565c0"
ORANGE = "#ff8c00"
GREEN  = "#00e676"
RED    = "#ff4444"
PURPLE = "#7c4dff"
TEAL   = "#1de9b6"
GOLD   = "#ffd740"
PAL    = [CYAN, ORANGE, GREEN, RED, PURPLE, TEAL, GOLD, BLUE]

plt.rcParams.update({
    "figure.facecolor":  "#0a0a1a",
    "axes.facecolor":    "#0d1b2a",
    "axes.edgecolor":    "#1a3a5c",
    "axes.labelcolor":   "#8ab4c8",
    "axes.titlecolor":   "#e8f4f8",
    "axes.titlesize":    12,
    "axes.titleweight":  "bold",
    "xtick.color":       "#4a7a9b",
    "ytick.color":       "#4a7a9b",
    "grid.color":        "#0d2137",
    "grid.linewidth":    0.7,
    "text.color":        "#8ab4c8",
    "legend.facecolor":  "#0d1b2a",
    "legend.edgecolor":  "#1a3a5c",
    "legend.fontsize":   9,
    "font.family":       "monospace",
    "figure.dpi":        110,
    "savefig.facecolor": "#0a0a1a",
})

import sklearn
print(f"✓  numpy      {np.__version__}")
print(f"✓  pandas     {pd.__version__}")
print(f"✓  sklearn    {sklearn.__version__}")
print("\n   Todo listo para comenzar.")
```

Cada dataset devuelve un objeto tipo <code>Bunch</code>. Sus atributos más útiles son <code>.data</code> (features), <code>.target</code> (etiquetas) y <code>.feature_names</code> (nombres de columnas).

```python
# ── Cargar los datasets más usados ───────────────────────────────────
iris    = datasets.load_iris()           # flores — clasificación (3 clases)
cancer  = datasets.load_breast_cancer()  # cáncer — clasificación binaria
wine    = datasets.load_wine()           # vinos  — clasificación (3 clases)
digits  = datasets.load_digits()         # dígitos — clasificación (10 clases)
housing = datasets.fetch_california_housing()  # viviendas — regresión

print("DATASETS DISPONIBLES")
print("─" * 58)
for nombre, ds in [("Iris",         iris),
                   ("Cáncer mama",  cancer),
                   ("Vinos",        wine),
                   ("Dígitos",      digits),
                   ("CA Housing",   housing)]:
    tipo = "clasificación" if hasattr(ds, "target_names") else "regresión"
    print(f"  {nombre:<16} {ds.data.shape[0]:>5} muestras  "
          f"{ds.data.shape[1]:>3} features  [{tipo}]")
```

```python
# ── Exploración rápida del dataset Iris ─────────────────────────────

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["especie"] = [iris.target_names[t] for t in iris.target]

print("Primeras filas:")
print(df.head())
print("\nEstadísticas básicas:")
print(df.describe().round(2))
```

```python
# ── Visualizar distribución de features por clase ────────────────────

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle("Dataset Iris — Distribución de features por especie",
             color="#e8f4f8", fontsize=12, y=1.02)

pares = [
    ("sepal length (cm)", "sepal width (cm)"),
    ("petal length (cm)", "petal width (cm)"),
    ("sepal length (cm)", "petal length (cm)"),
]

for ax, (xf, yf) in zip(axes, pares):
    for i, sp in enumerate(iris.target_names):
        sub = df[df["especie"] == sp]
        ax.scatter(sub[xf], sub[yf], color=PAL[i], s=45,
                   alpha=0.85, label=sp, edgecolors="none")
    ax.set_xlabel(xf.replace(" (cm)", ""), fontsize=9)
    ax.set_ylabel(yf.replace(" (cm)", ""), fontsize=9)
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.25)

plt.tight_layout()
plt.show()

print("\nDato clave: petal length y petal width separan muy bien las 3 especies.")
```

<strong style="color:#ff8c00;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">⚠️ Atención</strong>
Regla de oro: <strong>ajusta el scaler solo con datos de entrenamiento</strong>, luego aplica la misma transformación al conjunto de prueba. Si ajustas con todo el dataset, introduces fuga de información.
</div>

```python
# ── División entrenamiento / prueba ──────────────────────────────────
X = cancer.data
y = cancer.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,    # 25% para prueba
    random_state=42,   # semilla para reproducibilidad
    stratify=y         # misma proporción de clases en ambos conjuntos
)

print(f"Total de muestras:     {len(X)}")
print(f"Entrenamiento:         {len(X_train)}")
print(f"Prueba:                {len(X_test)}")
print(f"\nProporción clase positiva (train): {y_train.mean():.1%}")
print(f"Proporción clase positiva (test):  {y_test.mean():.1%}")
print("  → stratify=y garantiza que ambas proporciones sean iguales")
```

```python
# ── Comparar los tres scalers ────────────────────────────────────────

X_demo = housing.data[:, :4]
nombres_feat = [n.replace("Median", "Med.") for n in housing.feature_names[:4]]

scalers = {
    "StandardScaler\n(media=0, std=1)":  StandardScaler(),
    "MinMaxScaler\n(rango [0, 1])":       MinMaxScaler(),
    "RobustScaler\n(usa IQR, tolera outliers)": RobustScaler(),
}

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle("Comparación de scalers — CA Housing (4 features)",
             color="#e8f4f8", fontsize=12)

for ax, (label, scaler) in zip(axes, scalers.items()):
    X_scaled = scaler.fit_transform(X_demo)
    for i, name in enumerate(nombres_feat):
        ax.hist(X_scaled[:, i], bins=40, alpha=0.5,
                color=PAL[i], label=name, density=True)
    ax.set_title(label, fontsize=9, color=CYAN)
    ax.set_xlabel("Valor escalado", fontsize=8)
    ax.legend(fontsize=7)
    ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()

print("¿Cuándo usar cada uno?")
print("  StandardScaler → la mayoría de modelos lineales y SVM")
print("  MinMaxScaler   → redes neuronales, algoritmos basados en distancia")
print("  RobustScaler   → cuando hay outliers importantes en los datos")
```

```python
# ── Imputación de valores nulos ─────────────────────────────────────

rng   = np.random.default_rng(42)
X_nan = iris.data.copy().astype(float)
mask  = rng.random(X_nan.shape) < 0.15  # 15% de valores faltantes
X_nan[mask] = np.nan

print(f"Valores faltantes introducidos: {mask.sum()} de {X_nan.size} ({mask.mean():.1%})")
print()

for estrategia in ["mean", "median", "most_frequent"]:
    imp   = SimpleImputer(strategy=estrategia)
    X_imp = imp.fit_transform(X_nan)
    print(f"  strategy='{estrategia:<15}' → NaN restantes: {np.isnan(X_imp).sum()}")
```

<div style="background:#0d1b2a;border-left:3px solid #00d4ff;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:system-ui,sans-serif;font-size:13px;color:#8ab4c8;line-height:1.6">
<strong style="color:#00d4ff;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">💡 Concepto clave</strong>
El flujo siempre es el mismo: <code>fit(X_train, y_train)</code> → <code>predict(X_test)</code> → evaluar con métricas. Lo que cambia es el algoritmo.
</div>

```python
# ── Preparar datos ───────────────────────────────────────────────────

scaler_c = StandardScaler()
X_train_s = scaler_c.fit_transform(X_train)
X_test_s  = scaler_c.transform(X_test)

# ── Definir los clasificadores a comparar ────────────────────────────
clasificadores = {
    "Regresión Logística":   LogisticRegression(max_iter=1000, random_state=42),
    "Árbol de Decisión":     DecisionTreeClassifier(max_depth=5, random_state=42),
    "Random Forest":         RandomForestClassifier(n_estimators=200, random_state=42),
    "SVM (kernel RBF)":      SVC(kernel="rbf", probability=True, random_state=42),
    "KNN (k=7)":             KNeighborsClassifier(n_neighbors=7),
    "Gradient Boosting":     GradientBoostingClassifier(n_estimators=100, random_state=42),
    "Naïve Bayes":           GaussianNB(),
}

# ── Entrenar y evaluar ────────────────────────────────────────────────
resultados = {}
for nombre, clf in clasificadores.items():
    clf.fit(X_train_s, y_train)
    y_pred = clf.predict(X_test_s)
    y_prob = clf.predict_proba(X_test_s)[:, 1] if hasattr(clf, "predict_proba")              else clf.decision_function(X_test_s)
    acc  = accuracy_score(y_test, y_pred)
    auc  = roc_auc_score(y_test, y_prob)
    cv   = cross_val_score(clf, X_train_s, y_train, cv=5, scoring="accuracy").mean()
    resultados[nombre] = {"Accuracy": acc, "AUC-ROC": auc, "CV-5": cv}
    print(f"  {nombre:<24}  Acc={acc:.3f}  AUC={auc:.3f}  CV={cv:.3f}")

df_clf = pd.DataFrame(resultados).T.sort_values("AUC-ROC", ascending=False)
```

```python
# ── Gráfica comparativa de clasificadores ────────────────────────────

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("Comparación de clasificadores — Dataset Cáncer de Mama",
             color="#e8f4f8", fontsize=12)

# --- Barras de accuracy ---
nombres = list(df_clf.index)
accs    = df_clf["Accuracy"].values
colores = [CYAN if v == accs.max() else "#1a3a5c" for v in accs]
barras  = ax1.barh(nombres, accs, color=colores, height=0.55, edgecolor="none")
ax1.set_xlim(0.88, 1.01)
ax1.set_xlabel("Accuracy (datos de prueba)")
ax1.set_title("Accuracy por algoritmo", color=CYAN)
ax1.grid(True, alpha=0.2, axis="x")
for bar, v in zip(barras, accs):
    ax1.text(v + 0.001, bar.get_y() + bar.get_height() / 2,
             f"{v:.3f}", va="center", color="#e8f4f8", fontsize=9)

# --- Curvas ROC ---
for i, (nombre, clf) in enumerate(clasificadores.items()):
    if hasattr(clf, "predict_proba"):
        prob = clf.predict_proba(X_test_s)[:, 1]
    else:
        prob = clf.decision_function(X_test_s)
    fpr, tpr, _ = roc_curve(y_test, prob)
    auc = roc_auc_score(y_test, prob)
    ax2.plot(fpr, tpr, color=PAL[i % len(PAL)], lw=1.8,
             label=f"{nombre} ({auc:.3f})", alpha=0.9)

ax2.plot([0, 1], [0, 1], color="#2a4a6a", ls="--", lw=1.2, label="Aleatorio")
ax2.set_xlabel("Tasa de falsos positivos")
ax2.set_ylabel("Tasa de verdaderos positivos")
ax2.set_title("Curvas ROC (AUC)", color=CYAN)
ax2.legend(fontsize=8, loc="lower right")
ax2.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
# ── Matriz de confusión y reporte detallado ──────────────────────────

mejor_clf = clasificadores["Random Forest"]
y_pred_rf = mejor_clf.predict(X_test_s)
cm        = confusion_matrix(y_test, y_pred_rf)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Random Forest — Análisis detallado",
             color="#e8f4f8", fontsize=12)

# Matriz de confusión
im = ax1.imshow(cm, cmap="Blues", aspect="auto")
ax1.set_xticks([0, 1]); ax1.set_yticks([0, 1])
ax1.set_xticklabels(["Pred: Maligno", "Pred: Benigno"], fontsize=9)
ax1.set_yticklabels(["Real: Maligno", "Real: Benigno"], fontsize=9)
ax1.set_title("Matriz de confusión", color=CYAN)
plt.colorbar(im, ax=ax1)
for i in range(2):
    for j in range(2):
        ax1.text(j, i, str(cm[i, j]), ha="center", va="center",
                 color="#0a0a1a" if cm[i, j] > cm.max() / 2 else "#e8f4f8",
                 fontsize=20, fontweight="bold")

# Curva Precisión-Recall
prob_rf = mejor_clf.predict_proba(X_test_s)[:, 1]
prec, rec, _ = precision_recall_curve(y_test, prob_rf)
ax2.plot(rec, prec, color=CYAN, lw=2.5)
ax2.fill_between(rec, prec, alpha=0.1, color=CYAN)
ax2.axhline(y_test.mean(), color=RED, ls="--", lw=1.5,
            label=f"Baseline ({y_test.mean():.2f})")
ax2.set_xlabel("Recall")
ax2.set_ylabel("Precisión")
ax2.set_title("Curva Precisión-Recall", color=CYAN)
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()

print("\nReporte completo:")
print(classification_report(y_test, y_pred_rf,
      target_names=["Maligno", "Benigno"]))
```

Para regresión, las métricas clave son: <strong>RMSE</strong> (error en las mismas unidades que el target), <strong>MAE</strong> (error absoluto medio) y <strong>R²</strong> (qué tan bien explica el modelo la varianza, de 0 a 1).
</div>

```python
# ── Datos: California Housing ─────────────────────────────────────────
X_r, y_r = housing.data, housing.target

X_tr_r, X_te_r, y_tr_r, y_te_r = train_test_split(
    X_r, y_r, test_size=0.2, random_state=42
)
scaler_r = StandardScaler()
X_tr_rs  = scaler_r.fit_transform(X_tr_r)
X_te_rs  = scaler_r.transform(X_te_r)

# ── Comparar regresores ───────────────────────────────────────────────
regresores = {
    "Regresión Lineal":  LinearRegression(),
    "Ridge (L2)":        Ridge(alpha=1.0),
    "Lasso (L1)":        Lasso(alpha=0.01, max_iter=5000),
    "Random Forest":     RandomForestRegressor(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, random_state=42),
}

res_reg = {}
for nombre, reg in regresores.items():
    reg.fit(X_tr_rs, y_tr_r)
    y_hat = reg.predict(X_te_rs)
    rmse  = np.sqrt(mean_squared_error(y_te_r, y_hat))
    r2    = r2_score(y_te_r, y_hat)
    res_reg[nombre] = {"RMSE": rmse, "R²": r2}
    print(f"  {nombre:<22}  RMSE={rmse:.4f}  R²={r2:.4f}")

df_reg = pd.DataFrame(res_reg).T.sort_values("R²", ascending=False)
```

```python
# ── Gráficas: predicho vs real, residuos y R² comparado ──────────────

best_reg  = regresores["Gradient Boosting"]
y_hat_gbm = best_reg.predict(X_te_rs)
residuos  = y_te_r - y_hat_gbm

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Análisis de regresión — CA Housing (precio en $100k)",
             color="#e8f4f8", fontsize=12)

# Predicho vs real
ax = axes[0]
ax.scatter(y_te_r, y_hat_gbm, alpha=0.25, s=8,
           color=CYAN, edgecolors="none")
lims = [min(y_te_r.min(), y_hat_gbm.min()),
        max(y_te_r.max(), y_hat_gbm.max())]
ax.plot(lims, lims, color=RED, lw=1.5, ls="--", label="Predicción perfecta")
ax.set_xlabel("Valor real"); ax.set_ylabel("Valor predicho")
ax.set_title(f"Predicho vs Real  (R²={r2_score(y_te_r, y_hat_gbm):.4f})",
             color=CYAN)
ax.legend(fontsize=8); ax.grid(True, alpha=0.2)

# Residuos
ax = axes[1]
ax.scatter(y_hat_gbm, residuos, alpha=0.2, s=8,
           color=ORANGE, edgecolors="none")
ax.axhline(0, color=RED, lw=1.5, ls="--")
ax.set_xlabel("Valor predicho"); ax.set_ylabel("Residuo")
ax.set_title("Gráfico de residuos", color=CYAN)
ax.grid(True, alpha=0.2)

# R² comparado
ax = axes[2]
r2_vals  = df_reg["R²"].values
noms_r2  = list(df_reg.index)
cols_r2  = [GREEN if v == r2_vals.max() else "#1a3a5c" for v in r2_vals]
ax.barh(noms_r2, r2_vals, color=cols_r2, height=0.55, edgecolor="none")
ax.set_xlim(0, 1.05)
ax.set_xlabel("R²"); ax.set_title("Comparación R²", color=CYAN)
ax.grid(True, alpha=0.2, axis="x")
for i, v in enumerate(r2_vals):
    ax.text(v + 0.01, i, f"{v:.3f}", va="center",
            color="#e8f4f8", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# ── Regularización: Ridge vs Lasso — qué pasa con los coeficientes ──

alphas = np.logspace(-3, 3, 60)
ridge_coefs, lasso_coefs = [], []

for a in alphas:
    ridge_coefs.append(Ridge(alpha=a).fit(X_tr_rs, y_tr_r).coef_)
    lasso_coefs.append(Lasso(alpha=a, max_iter=10000).fit(X_tr_rs, y_tr_r).coef_)

ridge_coefs = np.array(ridge_coefs)
lasso_coefs = np.array(lasso_coefs)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
fig.suptitle("Regularización — Cómo cambian los coeficientes según alpha",
             color="#e8f4f8", fontsize=12)

for i, nombre in enumerate(housing.feature_names):
    ax1.plot(alphas, ridge_coefs[:, i], color=PAL[i % len(PAL)],
             lw=1.6, label=nombre)
    ax2.plot(alphas, lasso_coefs[:, i], color=PAL[i % len(PAL)],
             lw=1.6, label=nombre)

for ax, titulo in [(ax1, "Ridge L2 — coeficientes se encogen gradualmente"),
                   (ax2, "Lasso L1 — coeficientes llegan exactamente a cero")]:
    ax.set_xscale("log")
    ax.set_xlabel("Alpha →  más regularización")
    ax.set_ylabel("Valor del coeficiente")
    ax.set_title(titulo, color=CYAN, fontsize=10)
    ax.legend(fontsize=7, loc="upper right")
    ax.axhline(0, color="#2a4a6a", lw=1, ls="--")
    ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
print("Ridge: los coeficientes se acercan a 0 pero nunca llegan → mantiene todas las features")
print("Lasso: fuerza coeficientes exactamente a 0 → selección automática de features")
```

<strong style="color:#00d4ff;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">💡 Concepto clave</strong>
En clustering <strong>no tenemos etiquetas</strong>. El algoritmo descubre la estructura por sí solo. Para evaluar la calidad usamos el <strong>Silhouette Score</strong> (varía de -1 a 1; más alto es mejor) y el <strong>método del codo</strong>.

```python
# ── Método del codo + Silhouette para elegir K ───────────────────────

from sklearn.datasets import make_blobs, make_moons

X_cl  = StandardScaler().fit_transform(iris.data)
inercias, silhouettes = [], []
K_rango = range(2, 11)

for k in K_rango:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    etiquetas = km.fit_predict(X_cl)
    inercias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_cl, etiquetas))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("¿Cuántos clusters elegir? — Dataset Iris",
             color="#e8f4f8", fontsize=12)

# Método del codo
ax1.plot(list(K_rango), inercias, color=CYAN, marker="o", ms=8, lw=2.2)
ax1.axvline(3, color=RED, ls="--", lw=1.5, label="K=3 (codo)")
ax1.set_xlabel("Número de clusters (K)")
ax1.set_ylabel("Inercia (WCSS)")
ax1.set_title("Método del Codo", color=CYAN)
ax1.legend(); ax1.grid(True, alpha=0.2)
ax1.annotate("← codo aquí", xy=(3, inercias[1]),
             xytext=(5, inercias[3]),
             arrowprops=dict(arrowstyle="->", color="#8ab4c8"),
             color="#8ab4c8", fontsize=9)

# Silhouette
mejor_k = K_rango[int(np.argmax(silhouettes))]
ax2.plot(list(K_rango), silhouettes, color=GREEN, marker="s", ms=8, lw=2.2)
ax2.axvline(mejor_k, color=RED, ls="--", lw=1.5,
            label=f"Mejor K={mejor_k}  ({max(silhouettes):.3f})")
ax2.set_xlabel("Número de clusters (K)")
ax2.set_ylabel("Silhouette Score")
ax2.set_title("Silhouette Score", color=CYAN)
ax2.legend(); ax2.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
print("Regla práctica: busca el punto donde la inercia deja de bajar rápido (el 'codo')")
print("y confirma con Silhouette. Aquí K=3 tiene sentido porque hay 3 especies.")
```

```python
# ── Comparar K-Means, DBSCAN y Gaussian Mixture ──────────────────────

X_blobs, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)
X_moons, _ = make_moons(n_samples=300, noise=0.08, random_state=42)
X_blobs_s  = StandardScaler().fit_transform(X_blobs)
X_moons_s  = StandardScaler().fit_transform(X_moons)

algoritmos = {
    """K-Means
(bueno para grupos esféricos)""":       (KMeans(n_clusters=4, random_state=42, n_init=10), X_blobs_s),
    """DBSCAN
(detecta formas arbitrarias)""":         (DBSCAN(eps=0.3, min_samples=5), X_moons_s),
    """Gaussian Mixture
(fronteras suaves)""":         (GaussianMixture(n_components=4, random_state=42), X_blobs_s),
}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Comparación de algoritmos de clustering",
             color="#e8f4f8", fontsize=12)

for ax, (nombre, (algo, X_data)) in zip(axes, algoritmos.items()):
    etiquetas = algo.fit_predict(X_data)
    unicas    = np.unique(etiquetas)
    for lbl in unicas:
        mask  = etiquetas == lbl
        color = "#2a4a6a" if lbl == -1 else PAL[lbl % len(PAL)]
        label = "Ruido" if lbl == -1 else f"Grupo {lbl}"
        ax.scatter(X_data[mask, 0], X_data[mask, 1],
                   color=color, s=28, alpha=0.85,
                   edgecolors="none", label=label)
    sil = silhouette_score(X_data, etiquetas)           if len(unicas) > 1 and -1 not in unicas else float("nan")
    sil_txt = f"Silhouette={sil:.3f}" if not np.isnan(sil) else "(incluye ruido)"
    ax.set_title(f"""{nombre}
{sil_txt}""", color=CYAN, fontsize=9)
    ax.legend(fontsize=7)
    ax.grid(True, alpha=0.2)
    ax.set_xticks([]); ax.set_yticks([])

plt.tight_layout()
plt.show()

print("K-Means:         rápido, escalable, asume grupos del mismo tamaño")
print("DBSCAN:          detecta formas complejas, identifica ruido/outliers")
print("Gaussian Mixture: asigna probabilidades de pertenencia (clusters 'blandos')")
```

```python
# ── PCA en dataset Digits (64 features → 2D) ─────────────────────────

X_d   = StandardScaler().fit_transform(digits.data)
y_d   = digits.target

pca_full = PCA(random_state=42).fit(X_d)
var_cum  = np.cumsum(pca_full.explained_variance_ratio_)
n_para_95 = np.searchsorted(var_cum, 0.95) + 1

pca_2d   = PCA(n_components=2, random_state=42)
X_pca_2d = pca_2d.fit_transform(X_d)

tsne    = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=500, verbose=0)
X_tsne  = tsne.fit_transform(X_d[:1000])
y_tsne  = y_d[:1000]

fig = plt.figure(figsize=(20, 5))
gs  = gridspec.GridSpec(1, 4, figure=fig, wspace=0.3)
fig.suptitle("Reducción Dimensional — Dataset Dígitos (64 features originales)",
             color="#e8f4f8", fontsize=12)

# Scree plot
ax1 = fig.add_subplot(gs[0])
ax1.bar(range(1, 21), pca_full.explained_variance_ratio_[:20] * 100,
        color=CYAN, alpha=0.85, edgecolor="none")
ax1.set_xlabel("Componente principal")
ax1.set_ylabel("Varianza explicada (%)")
ax1.set_title("Scree Plot (top 20)", color=CYAN)
ax1.grid(True, alpha=0.2, axis="y")

# Varianza acumulada
ax2 = fig.add_subplot(gs[1])
ax2.plot(range(1, len(var_cum) + 1), var_cum * 100, color=GREEN, lw=2)
ax2.axhline(95, color=RED, ls="--", lw=1.2, label="95%")
ax2.axhline(80, color=ORANGE, ls="--", lw=1.2, label="80%")
ax2.axvline(n_para_95, color=RED, ls=":", lw=1)
ax2.set_xlabel("N° de componentes")
ax2.set_ylabel("Varianza acumulada (%)")
ax2.set_title(f"""Varianza acumulada
(95% con solo {n_para_95} componentes)""", color=CYAN)
ax2.legend(fontsize=8); ax2.grid(True, alpha=0.2)

# PCA 2D
ax3 = fig.add_subplot(gs[2])
for digito in range(10):
    mask = y_d == digito
    ax3.scatter(X_pca_2d[mask, 0], X_pca_2d[mask, 1],
                s=12, alpha=0.6, edgecolors="none",
                color=PAL[digito % len(PAL)], label=str(digito))
var_pct = pca_2d.explained_variance_ratio_.sum()
ax3.set_title(f"""PCA 2D
({var_pct:.1%} de varianza explicada)""", color=CYAN)
ax3.legend(ncol=2, fontsize=7, markerscale=2)
ax3.grid(True, alpha=0.2)
ax3.set_xticks([]); ax3.set_yticks([])

# t-SNE 2D
ax4 = fig.add_subplot(gs[3])
for digito in range(10):
    mask = y_tsne == digito
    ax4.scatter(X_tsne[mask, 0], X_tsne[mask, 1],
                s=12, alpha=0.6, edgecolors="none",
                color=PAL[digito % len(PAL)], label=str(digito))
ax4.set_title(f"""t-SNE 2D (1.000 muestras)
⚠️ Solo para visualización""", color=CYAN)
ax4.legend(ncol=2, fontsize=7, markerscale=2)
ax4.grid(True, alpha=0.2)
ax4.set_xticks([]); ax4.set_yticks([])

plt.savefig("dim_reduction.png", dpi=110, bbox_inches="tight")
plt.show()

print(f"Con solo {n_para_95} componentes (de 64) se captura el 95% de la información.")
print("t-SNE separa mejor los clusters visualmente, pero sus ejes no tienen interpretación.")
```

<strong style="color:#00e676;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">✅ Buena práctica</strong>
Un Pipeline evita la fuga de información automáticamente: el scaler se ajusta solo con datos de entrenamiento. Además, puedes hacer GridSearch directamente sobre el pipeline completo.
</div>

```python
# ── Pipeline básico ────────────────────────────────────────────────────

pipeline = Pipeline([
    ("escalador", StandardScaler()),
    ("pca",       PCA(n_components=15)),
    ("modelo",    RandomForestClassifier(n_estimators=100, random_state=42)),
])

# Todo en una sola línea — no hay riesgo de fuga
pipeline.fit(X_train, y_train)
score = pipeline.score(X_test, y_test)

print(f"Accuracy del Pipeline: {score:.4f}")
print()
print("Pasos del pipeline:")
for nombre, paso in pipeline.steps:
    print(f"  {nombre:<12} → {type(paso).__name__}")
```

```python
# ── GridSearchCV sobre el pipeline completo ──────────────────────────

param_grid = {
    "pca__n_components":   [5, 10, 20],
    "modelo__n_estimators": [50, 100, 200],
    "modelo__max_depth":    [None, 5, 10],
}

busqueda = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="roc_auc",
    n_jobs=-1,
    verbose=0,
)

print("Buscando los mejores hiperparámetros... (puede tardar ~30 segundos)")
busqueda.fit(X_train, y_train)

print(f"\nMejores parámetros: {busqueda.best_params_}")
print(f"Mejor AUC en CV:     {busqueda.best_score_:.4f}")
print(f"AUC en test:         {roc_auc_score(y_test, busqueda.predict_proba(X_test)[:, 1]):.4f}")
```

```python
# ── Importancia desde Random Forest ──────────────────────────────────

rf_imp = RandomForestClassifier(n_estimators=300, random_state=42)
rf_imp.fit(X_train_s, y_train)

importancias = rf_imp.feature_importances_
nombres_feat = cancer.feature_names
orden        = np.argsort(importancias)[::-1]

# ── Comparar con Mutual Information y RFE ─────────────────────────────
from sklearn.feature_selection import mutual_info_classif

mi = mutual_info_classif(X_train_s, y_train, random_state=42)
rfe = RFE(LogisticRegression(max_iter=2000), n_features_to_select=10)
rfe.fit(X_train_s, y_train)

top_n = 15
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("¿Qué features importan más? — Cáncer de Mama",
             color="#e8f4f8", fontsize=12)

# Random Forest
idx = orden[:top_n]
ax1.barh(nombres_feat[idx][::-1], importancias[idx][::-1],
         color=CYAN, alpha=0.85, edgecolor="none")
ax1.set_title("Importancia (Random Forest — Gini)", color=CYAN)
ax1.set_xlabel("Importancia relativa")
ax1.grid(True, alpha=0.2, axis="x")

# Mutual Information
mi_orden = np.argsort(mi)[::-1][:top_n]
ax2.barh(nombres_feat[mi_orden][::-1], mi[mi_orden][::-1],
         color=ORANGE, alpha=0.85, edgecolor="none")
ax2.set_title("Información Mutua", color=CYAN)
ax2.set_xlabel("Score de información mutua")
ax2.grid(True, alpha=0.2, axis="x")

plt.tight_layout()
plt.show()

top_rf = set(nombres_feat[orden[:10]])
top_mi = set(nombres_feat[mi_orden[:10]])
top_rfe = set(nombres_feat[rfe.support_])
comun   = top_rf & top_mi & top_rfe
print(f"Features en top-10 de los 3 métodos: {comun}")
```

<strong style="color:#00d4ff;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">💡 Concepto clave</strong>
Si la curva de entrenamiento es mucho mejor que la de validación → <strong>overfitting</strong>. Si ambas son malas → <strong>underfitting</strong>. El objetivo es que converjan a un valor alto.
</div>

```python
# ── Curvas de aprendizaje ─────────────────────────────────────────────

pipe_lc = Pipeline([
    ("scaler", StandardScaler()),
    ("modelo", RandomForestClassifier(n_estimators=100, random_state=42)),
])

tam_train, scores_train, scores_val = learning_curve(
    pipe_lc, cancer.data, cancer.target,
    train_sizes=np.linspace(0.1, 1.0, 10),
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    scoring="accuracy",
    n_jobs=-1,
)

# ── Validation curve: efecto de la profundidad máxima ─────────────────
from sklearn.model_selection import validation_curve

pipe_vc    = Pipeline([
    ("scaler", StandardScaler()),
    ("modelo", RandomForestClassifier(n_estimators=50, random_state=42)),
])
param_rango = [1, 2, 3, 5, 8, 10, 15, 20, None]
vc_train, vc_val = validation_curve(
    pipe_vc, cancer.data, cancer.target,
    param_name="modelo__max_depth",
    param_range=param_rango,
    cv=5, scoring="accuracy", n_jobs=-1,
)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle("Diagnóstico del modelo — ¿Overfitting o Underfitting?",
             color="#e8f4f8", fontsize=12)

# Curva de aprendizaje
tr_m = scores_train.mean(axis=1)
tr_s = scores_train.std(axis=1)
va_m = scores_val.mean(axis=1)
va_s = scores_val.std(axis=1)

ax1.plot(tam_train, tr_m, color=CYAN, lw=2.2, marker="o", ms=6, label="Entrenamiento")
ax1.fill_between(tam_train, tr_m - tr_s, tr_m + tr_s, alpha=0.12, color=CYAN)
ax1.plot(tam_train, va_m, color=ORANGE, lw=2.2, marker="s", ms=6, label="Validación CV")
ax1.fill_between(tam_train, va_m - va_s, va_m + va_s, alpha=0.12, color=ORANGE)
ax1.set_xlabel("Tamaño del set de entrenamiento")
ax1.set_ylabel("Accuracy")
ax1.set_title(f"""Curva de Aprendizaje
¿Más datos mejoran el modelo?""", color=CYAN)
ax1.legend(); ax1.grid(True, alpha=0.2)

# Validation curve
x_lab = [str(d) if d is not None else "None" for d in param_rango]
ax2.plot(x_lab, vc_train.mean(axis=1), color=CYAN, lw=2.2, marker="o", ms=6, label="Entrenamiento")
ax2.fill_between(x_lab,
                 vc_train.mean(axis=1) - vc_train.std(axis=1),
                 vc_train.mean(axis=1) + vc_train.std(axis=1),
                 alpha=0.12, color=CYAN)
ax2.plot(x_lab, vc_val.mean(axis=1), color=ORANGE, lw=2.2, marker="s", ms=6, label="Validación CV")
ax2.fill_between(x_lab,
                 vc_val.mean(axis=1) - vc_val.std(axis=1),
                 vc_val.mean(axis=1) + vc_val.std(axis=1),
                 alpha=0.12, color=ORANGE)
ax2.set_xlabel("Profundidad máxima del árbol (max_depth)")
ax2.set_ylabel("Accuracy")
ax2.set_title(f"""Validation Curve
Efecto de max_depth en Random Forest""", color=CYAN)
ax2.legend(); ax2.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()

brecha = tr_m[-1] - va_m[-1]
print(f"Brecha final Entrenamiento - Validación: {brecha:.4f}")
if brecha < 0.02:
    print("  → Buena generalización, bajo overfitting")
elif brecha > 0.1:
    print("  → Posible overfitting — probar más regularización o menos features")
else:
    print("  → Brecha normal para este tipo de modelo")
```

<strong style="color:#00d4ff;font-size:11px;letter-spacing:2px;text-transform:uppercase;display:block;margin-bottom:4px">💡 Concepto clave</strong>
La intuición detrás de los ensembles: combinar muchos modelos 'mediocres' produce un modelo 'excelente'. <strong>Bagging</strong> reduce la varianza, <strong>Boosting</strong> reduce el bias. <strong>Voting</strong> y <strong>Stacking</strong> combinan los mejores de ambos mundos.
</div>

```python
from sklearn.ensemble import BaggingClassifier, StackingClassifier

# Modelos base
lr  = LogisticRegression(max_iter=1000, random_state=42)
dt  = DecisionTreeClassifier(max_depth=5, random_state=42)
knn = KNeighborsClassifier(n_neighbors=7)
nb  = GaussianNB()
rf  = RandomForestClassifier(n_estimators=100, random_state=42)
gb  = GradientBoostingClassifier(n_estimators=100, random_state=42)
ada = AdaBoostClassifier(n_estimators=100, random_state=42, algorithm="SAMME")

# Métodos ensemble
ensembles = {
    "Solo: Árbol de Decisión":  dt,
    "Solo: Reg. Logística":     lr,
    "Bagging (árboles)":        BaggingClassifier(DecisionTreeClassifier(), n_estimators=50, random_state=42),
    "AdaBoost":                 ada,
    "Gradient Boosting":        gb,
    "Random Forest":            rf,
    "Voting suave (LR+RF+GB)":  VotingClassifier([("lr",lr),("rf",rf),("gb",gb)], voting="soft"),
    "Stacking (LR+KNN+NB)":     StackingClassifier(
                                    estimators=[("lr",lr),("knn",knn),("nb",nb)],
                                    final_estimator=LogisticRegression(max_iter=1000),
                                    cv=5
                                ),
}

res_ens = {}
for nombre, clf in ensembles.items():
    clf.fit(X_train_s, y_train)
    y_pred = clf.predict(X_test_s)
    acc    = accuracy_score(y_test, y_pred)
    cv     = cross_val_score(clf, X_train_s, y_train, cv=5, scoring="accuracy").mean()
    res_ens[nombre] = {"Test": acc, "CV-5": cv}

df_ens = pd.DataFrame(res_ens).T.sort_values("Test", ascending=False)
print(df_ens.round(4).to_string())
```
